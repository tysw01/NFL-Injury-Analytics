```python
import pandas as pd
import re

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)

# --- Injury statement extraction ---
sentence_splitter = re.compile(r"(?<=[.!?])\s+")

start_patterns = [
    r"\binjury\s*update\b",
    r"\binjury\s*report\b",
    r"\binjur(?:ed|y|ies)\b",
    r"\bhas\s+an?\s+[a-z\-]+\s+injury\b",
    r"\bhas\s+an?\s+[a-z\-]+\s+concussion\b",
    r"\bhave\s+[a-z\-]+\s+injur(?:y|ies)\b",
    r"\bconcussion\b",
    r"\bwill\s+not\s+return\b",
    r"timeout\s+charged\s+to\s+team\s+due\s+to\s+injury",
]

continuation_patterns = [
    r"\breturn\b",
    r"\bquestionable\b",
    r"\bprobable\b",
    r"\bdoubtful\b",
    r"\bout\b",
    r"\bwill\s+not\s+return\b",
    r"\bbeing\s+evaluated\b",
    r"\blocker\s+room\b",
    r"\bcramp\b",
    r"\bfeeling\b",
    r"\bmotion\b",
    r"\binjury\b",
    r"\bconcussion\b",
    r"\bhospital\b",
    r"\bobservation\b",
    r"\bx[-\s]?rays?\b",
]

start_re = re.compile("|".join(start_patterns), flags=re.IGNORECASE)
cont_re = re.compile("|".join(continuation_patterns), flags=re.IGNORECASE)

name_pattern = r"(?:[A-Z]\.[A-Za-z']+|[A-Z][a-z]+)(?:\s[A-Z][a-z]+)*"
player_hint_patterns = [
    r"\b[A-Z]{2,3}-\d{1,3}-[A-Z]\.[A-Za-z']+\b",
    r"\b(?:No\.?\s*|#\s*)?\d{1,3}\s*-\s*" + name_pattern + r"\s*\([A-Z]{2,3}\)",
    r"\b" + name_pattern + r"\s+has\s+an?\s+[A-Za-z\-]+\s+injury\b",
    r"\b" + name_pattern + r"\s+was\s+injured\b",
    r"\b[A-Za-z]+\'s\s+" + name_pattern,
    r"\b" + name_pattern + r"\s+injured\b",
]
player_hint_re = re.compile("|".join(player_hint_patterns), flags=re.IGNORECASE)

def extract_injury_statements(desc):
    if not isinstance(desc, str) or not desc.strip():
        return []
    sentences = [s.strip() for s in sentence_splitter.split(desc) if s.strip()]
    results = []
    i = 0
    while i < len(sentences):
        sent = sentences[i]
        if start_re.search(sent):
            if results and not player_hint_re.search(sent) and not re.search(r"\binjury\s*update\b", sent, flags=re.IGNORECASE):
                results[-1] = f"{results[-1]} {sent}"
                i += 1
                continue
            current = [sent]
            j = i + 1
            while j < len(sentences):
                next_sent = sentences[j]
                if start_re.search(next_sent) and player_hint_re.search(next_sent):
                    break
                if cont_re.search(next_sent):
                    current.append(next_sent.strip())
                    j += 1
                    continue
                break
            results.append(" ".join(current))
            i = j
        else:
            i += 1
    return results

pbp_injury_statements = pbp.copy()
pbp_injury_statements["injury_statements"] = pbp_injury_statements["desc"].apply(extract_injury_statements)
pbp_injury_statements = pbp_injury_statements.explode("injury_statements", ignore_index=True)
pbp_injury_statements = pbp_injury_statements[pbp_injury_statements["injury_statements"].notna()]

pbp_injury_statements["has_player_hint"] = pbp_injury_statements["injury_statements"].apply(
    lambda x: bool(player_hint_re.search(x)) if isinstance(x, str) else False
)
timeout_mask = pbp_injury_statements["injury_statements"].str.contains(
    r"timeout\s+charged\s+to\s+team\s+due\s+to\s+injury",
    case=False,
    na=False,
)
group_has_player = pbp_injury_statements.groupby(["game_id", "play_id"])["has_player_hint"].transform("any")
pbp_injury_statements = pbp_injury_statements[~(timeout_mask & group_has_player)].copy()
pbp_injury_statements.drop(columns=["has_player_hint"], inplace=True)

pbp_injury_statements = pbp_injury_statements.drop_duplicates(subset=["game_id", "play_id", "injury_statements"])
pbp_injury_statements = pbp_injury_statements[
    ["game_id", "play_id", "game_date", "home_team", "away_team", "desc", "injury_statements"]
] .rename(columns={"injury_statements": "injury_text"})

# --- Parse player(s), number, team, status ---
strict_player_patterns = [
    re.compile(r"\b(?P<number>\d{1,3})\s+(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\s*\((?P<team>[A-Z]{2,3})\)\s+injured(?:\s+on\s+play)?\b", re.IGNORECASE),
    re.compile(r"\b#(?P<number>\d{1,3})\s+(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\s*\((?P<team>[A-Z]{2,3})\)\s+injured(?:\s+on\s+play)?\b", re.IGNORECASE),
    re.compile(r"\b(?P<team>[A-Z]{2,3})-(?P<number>\d{1,3})-(?P<name>[A-Z]\.[A-Za-z']+)\b"),
    re.compile(r"\b(?:No\.?\s*|#\s*)?(?P<number>\d{1,3})\s*-\s*(?P<name>" + name_pattern + r")\s*\((?P<team>[A-Z]{2,3})\)", re.IGNORECASE),
    re.compile(r"\b(?P<team>[A-Z]{2,3})\s*#\s*(?P<number>\d{1,3})\s+(?P<name>[A-Z]\.[A-Za-z']+|" + name_pattern + r")\b"),
    re.compile(r"\b(?:#\s*)?(?P<number>\d{1,3})\s+(?P<name>[A-Z]\.[A-Za-z']+)\s+injured\b", re.IGNORECASE),
    re.compile(r"\b(?P<team>[A-Za-z]{2,})'s\s+(?P<name>[A-Z]\.[A-Za-z']+|[A-Z][a-z]+(?:\s[A-Z][a-z]+)?)\b", re.IGNORECASE),
]

enrich_only_patterns = [
    re.compile(r"\b(?P<number>\d{1,3})\s+(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\s*\((?P<team>[A-Z]{2,3})\)\b", re.IGNORECASE),
    re.compile(r"\b(?P<team>[A-Z]{2,3})-(?P<number>\d{1,3})-(?P<name>[A-Za-z]{1,3}\.[A-Za-z']+)\b"),
    re.compile(r"\b(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\s+of\s+(?P<team>[A-Za-z]+)\b", re.IGNORECASE),
    re.compile(r"\binjury\s+updates?\s*:\s*(?:(?P<team>[A-Za-z]{2,})[’']s\s+)?(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\b", re.IGNORECASE),
    re.compile(r"\b#(?P<number>\d{1,3})\s+(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\b", re.IGNORECASE),
    re.compile(r"\b#(?P<number>\d{1,3})\s+(?P<team>[A-Za-z]{2,})\s+(?P<name>[A-Z][A-Za-z']+(?:\s+[A-Z][A-Za-z']+)*)\b", re.IGNORECASE),
    re.compile(r"\binjury\s+updates?\s*:\s*(?:(?P<team>[A-Za-z]{2,})[’']s\s+)?(?P<name>[A-Z][A-Za-z']+)\s+has\b", re.IGNORECASE),
    re.compile(r"\b(?P<name>[A-Z][A-Za-z']+)\s+has\s+[A-Za-z][A-Za-z\s'-]*\s+injury\b", re.IGNORECASE),
]

full_name_re = re.compile(r"\b(?P<name>[A-Z][a-z]+\s+[A-Z][a-z]+)\s+(?:has|was)\b")
dual_name_re = re.compile(
    r"\b(?P<name1>[A-Z][a-z]+\s+[A-Z][a-z]+)\s+and\s+(?P<name2>[A-Z][a-z]+\s+[A-Z][a-z]+)\s+(?:both\s+)?have\b",
    re.IGNORECASE,
)

def _injury_context_near_match(text, start, end):
    window = text[max(0, start - 90): min(len(text), end + 90)]
    return bool(re.search(
        r"\binjur(?:y|ed|ies)\b|\bconcussion\b|\bquestionable\b|\bprobable\b|\bdoubtful\b|\bout\b|\breturn\b|\bevaluated\b",
        window,
        flags=re.IGNORECASE
    ))

status_patterns = [
    re.compile(r"\b(questionable|probable|doubtful|out)\b", re.IGNORECASE),
    re.compile(r"\bwill\s+not\s+return\b", re.IGNORECASE),
]

def normalize_name(value):
    if not value:
        return None
    name = re.sub(r"\s+", " ", str(value)).strip(" .,:;-")
    name = re.sub(r"\s+(?:has|have|was|is|injured)\b.*$", "", name, flags=re.IGNORECASE)
    name = re.sub(r"\s+injury\b.*$", "", name, flags=re.IGNORECASE)
    return name or None

def infer_context_team(text):
    if not isinstance(text, str):
        return None
    m = re.search(r"INJURY\s+UPDATE\s*\((?P<team>[A-Z]{2,3})\)", text, flags=re.IGNORECASE)
    if m:
        return m.group("team").upper()
    m = re.search(r"INJURY\s+UPDATE\s*[-:]\s*(?P<team>[A-Z]{2,3})\s*#?\d", text, flags=re.IGNORECASE)
    if m:
        return m.group("team").upper()
    m = re.search(r"INJURY\s+(?:UPDATE|REPORT)\s+for\s+the\s+(?P<team>[A-Za-z]+)", text, flags=re.IGNORECASE)
    if m:
        return m.group("team")
    return None

def parse_player_entries(text):
    default = [{"player_name": None, "player_number": None, "player_team": None}]
    if not isinstance(text, str):
        return default

    context_team = infer_context_team(text)
    entries = []

    def upsert(name, number, team, allow_new):
        if not name:
            return False

        for existing in entries:
            if existing["player_name"] != name:
                continue

            number_conflict = (
                existing["player_number"] is not None and number is not None
                and existing["player_number"] != number
            )
            team_conflict = (
                existing["player_team"] is not None and team is not None
                and existing["player_team"] != team
            )
            if number_conflict or team_conflict:
                continue

            if existing["player_number"] is None and number is not None:
                existing["player_number"] = number
            if existing["player_team"] is None and team is not None:
                existing["player_team"] = team
            return True

        if allow_new:
            entries.append({"player_name": name, "player_number": number, "player_team": team})
            return True
        return False

    # strict pass can create rows
    for pat in strict_player_patterns:
        for m in pat.finditer(text):
            data = m.groupdict()
            name = normalize_name(data.get("name"))
            number = data.get("number")
            team = data.get("team") or context_team
            upsert(name, number, team, allow_new=True)

    # broad pass is fill-only
    for pat in enrich_only_patterns:
        for m in pat.finditer(text):
            if not _injury_context_near_match(text, m.start(), m.end()):
                continue
            data = m.groupdict()
            name = normalize_name(data.get("name"))
            number = data.get("number")
            team = data.get("team") or context_team

            filled = upsert(name, number, team, allow_new=False)
            if not filled and not entries and (number is not None or team is not None):
                upsert(name, number, team, allow_new=True)

    # name-only fallback only if nothing else matched
    if not entries:
        for m in dual_name_re.finditer(text):
            for field in ("name1", "name2"):
                name = normalize_name(m.group(field))
                upsert(name, None, context_team, allow_new=True)

        for m in full_name_re.finditer(text):
            name = normalize_name(m.group("name"))
            upsert(name, None, context_team, allow_new=True)

    return entries if entries else default

def parse_status(text):
    if not isinstance(text, str):
        return None
    for pat in status_patterns:
        m = pat.search(text)
        if m:
            return m.group(0)
    return None

pbp_injury_statements["player_entries"] = pbp_injury_statements["injury_text"].apply(parse_player_entries)
pbp_injury_statements = pbp_injury_statements.explode("player_entries", ignore_index=True)
player_df = pd.json_normalize(pbp_injury_statements["player_entries"])
pbp_injury_statements = pd.concat([pbp_injury_statements.drop(columns=["player_entries"]), player_df], axis=1)
pbp_injury_statements["return_status"] = pbp_injury_statements["injury_text"].apply(parse_status)

print(pbp_injury_statements.head(10000))
print(f"Total injury statements: {len(pbp_injury_statements)}")
```

```python
# Parser fix v2: fill missing player fields from timeout/injury-update variants
import re
import pandas as pd

def apply_parser_fix_v2(pbp_injury_statements: pd.DataFrame) -> pd.DataFrame:
    """Fill missing player fields from common timeout/injury-update patterns."""
    fallback_patterns = [
        re.compile(r"\b(?P<name>[A-Z][A-Za-z'\-]+(?:\s+[A-Z][A-Za-z'\-]+)*)\s*\(#(?P<number>\d{1,3})\)\s*(?:of\s+(?P<team>[A-Z]{2,3}))?\s+injured(?:\s+on\s+play)?\b", re.IGNORECASE),
        re.compile(r"\b(?P<name>[A-Z][A-Za-z'\-]+(?:\s+[A-Z][A-Za-z'\-]+)*)\s*\((?P<team>[A-Z]{2,3})\)\s+injured(?:\s+on\s+play)?\b", re.IGNORECASE),
        re.compile(r"\b(?:#\s*)?(?P<number>\d{1,3})\s+(?P<name>[A-Z]\.[A-Za-z'\-]+|[A-Z][A-Za-z'\-]+)\s+injured\b", re.IGNORECASE),
        re.compile(r"\binjury\s+to\s+(?P<team>[A-Z]{2,3})\s+(?P<name>[A-Z]\.[A-Za-z'\-]+|[A-Z][A-Za-z'\-]+(?:\s+[A-Z][A-Za-z'\-]+)?)\b", re.IGNORECASE),
        re.compile(r"\binjury\s*(?:timeout)?\s*(?:for)?\s*(?P<team>[A-Z]{2,3})\s*#?(?P<number>\d{1,3})\s*(?P<name>[A-Z]\.[A-Za-z'\-]+|[A-Z][A-Za-z'\-]+)?\b", re.IGNORECASE),
    ]

    def _fallback_entries(text: str) -> list[dict]:
        if not isinstance(text, str):
            return []
        out = []
        for pat in fallback_patterns:
            for m in pat.finditer(text):
                d = m.groupdict()
                name = (d.get("name") or "").strip(" .,:;-{}()")
                if not name:
                    continue
                out.append(
                    {
                        "player_name": name,
                        "player_number": d.get("number"),
                        "player_team": d.get("team"),
                    }
                )
        return out

    out_df = pbp_injury_statements.copy()
    mask_missing = out_df["player_name"].isna()

    for idx, row in out_df[mask_missing].iterrows():
        cands = _fallback_entries(row.get("injury_text"))
        if not cands:
            continue
        best = cands[0]
        out_df.at[idx, "player_name"] = best.get("player_name")
        if pd.isna(out_df.at[idx, "player_number"]) and best.get("player_number") is not None:
            out_df.at[idx, "player_number"] = best.get("player_number")
        if pd.isna(out_df.at[idx, "player_team"]) and best.get("player_team") is not None:
            out_df.at[idx, "player_team"] = best.get("player_team")

    return out_df

# Apply after your main parser has built pbp_injury_statements
before_missing = int(pbp_injury_statements["player_name"].isna().sum())
pbp_injury_statements = apply_parser_fix_v2(pbp_injury_statements)
after_missing = int(pbp_injury_statements["player_name"].isna().sum())

print(f"Parser fix applied. Missing player_name: {before_missing} -> {after_missing}")
```
